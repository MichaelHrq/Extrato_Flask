```python
import re
```

```python
frase =[
'26/04/2021 inss 263716 1.100,00 1.100,00 pacote de servicos padronizado prioritarios i 10421 -13,60 1.086,40',
'14/05/2021 pacote de servicos vr.parcial padronizado prior 30521 -6,40 0,00'
]
```

```python
filtro2 = []

for txt in frase:
    padraoData = re.compile(r'\d{2}/\d{2}/\d{4}')
    padrao = re.compile(r'\s?[a-z][0-9a-z\s\.]+[a-z0-9] [0-9]+ [0-9.,-]+ [0-9.,-]+\s?')
    data = re.findall(padraoData, txt)
    teste = re.findall(padrao, txt)
    for i in range(len(teste)):
        if (teste[i][0] == ' ') : filtro2.append(data[0]+teste[i])
        else : filtro2.append(data[0]+' '+teste[i])
```
